## Reproducibility

```python
# For reproducibility purposes

import random
import os

seed = 777

def set_seed(seed):
    random.seed(seed)                             # Python random module
    np.random.seed(seed)                          # NumPy
    os.environ['PYTHONHASHSEED'] = str(seed)      # Python hashing

set_seed(seed)
```
